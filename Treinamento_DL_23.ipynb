## NAS (Network Architecture Search)

Para entender o YOLO-NAS, é necessário explorar como a equipe de pesquisa da Deci descobriu a arquitetura de última geração. E isso envolve uma breve introdução ao conceito de NAS e AutoNAC.

NAS (Network Architecture Search) é um subcampo do AutoML (automated machine learning). NAS envolve o desenvolvimento de modelos de aprendizado de máquina que projetam e configuram automaticamente arquiteturas de redes neurais profundas com o objetivo de superar suas contrapartes projetadas, configuradas e construídas manualmente por pesquisadores humanos.

O avanço da pesquisa NAS facilita a descoberta eficiente de arquiteturas de redes neurais de alto desempenho por meio de um processo sistematizado que considera fatores como velocidade de inferência, recursos computacionais disponíveis, complexidade arquitetônica e precisão de previsão. Esforços bem-sucedidos de design e desenvolvimento de novas arquiteturas produzidas por meio de técnicas NAS ultrapassaram as arquiteturas de engenharia humana em tarefas como processamento de linguagem natural, detecção de objetos e classificação de imagens.

Um benefício significativo da utilização de técnicas NAS é que a simplificação dos processos de aprendizagem automática envolvidos na investigação, nomeadamente através da automação, reduz as horas humanas envolvidas nos esforços de investigação. Além disso, a escassez ou falta de conhecimento especializado no domínio é combatida pela introdução de sistemas que podem considerar múltiplos parâmetros para otimizar e explorar múltiplas configurações de rede interna em um vasto espaço de busca por meio de força bruta, que eventualmente configura a rede ideal com configurações que, em alguns casos, só pode ser alcançado com conhecimento humano especializado, criatividade e intuição.

As técnicas NAS são ainda mais relevantes hoje, pois se concentram na otimização de fatores como recursos computacionais, eficiência, precisão, consumo de energia e uso de memória, que são cruciais para adaptar arquiteturas a dispositivos de ponta (smartphones) ou cenários em tempo real.

Notavelmente, modelos projetados para serem executados em dispositivos de ponta e ambientes com escassez de computação, como MobileNetV3 e EfficientDet, foram descobertos projetando uma nova arquitetura neural e conduzindo uma busca automatizada por arquiteturas adequadas que melhorassem o desempenho de última geração em visão computacional tarefas.

NAS oferece benefícios significativos para pesquisadores. Mas a capacidade de percorrer um espaço de busca de possíveis arquiteturas e configurações de redes neurais profundas que consistem em milhões de parâmetros, combinados com fatores como recursos de computação, métricas de precisão de alvo e muito mais, requer extensos recursos de computação. Os grandes recursos computacionais e o conhecimento especializado necessários para executar técnicas NAS limitam sua utilização efetiva a um pequeno número de organizações.

## Automated Neural Architecture Construction (AutoNAC)

A construção automatizada de arquitetura neural (AutoNAC) é uma tecnologia NAS proprietária da Deci que explora com eficiência um vasto espaço de pesquisa de diversas configurações e estruturas arquitetônicas, considerando os tipos de blocos abrangentes, o número de blocos e as alocações de canais.

AutoNAC facilitou a descoberta da inovadora arquitetura YOLO-NAS e suas variantes (arquiteturas YOLO-NAS-S, YOLO-NAS-M e YOLO-NAS-L) procurando uma arquitetura de modelo ideal que combinasse as contribuições arquitetônicas fundamentais de Variantes do YOLO e incorporou vários componentes neurais inovadores da equipe de pesquisa da Deci que permitiram treinamento e inferência otimizados. Isso elevou o número possível de arquiteturas de redes neurais para 10¹⁴. Para colocar em perspectiva a vastidão do espaço de busca, isso é maior do que o número de estrelas em nossa galáxia.

O espaço de pesquisa que levou à descoberta da arquitetura YOLO-NAS incorporou princípios, design e considerações de computação que priorizaram eficiência, escalabilidade, robustez e interpretabilidade.

image.png

Eficiência refere-se ao requisito moderno de modelos de aprendizagem profunda para atender aos requisitos de armazenamento computacional de dispositivos de ponta, como smartphones, ao mesmo tempo que atende aos requisitos de desempenho em termos de velocidade e precisão de inferência. A arquitetura YOLO-NAS oferece desempenho de alta precisão para tarefas de detecção de objetos sem exigir grandes recursos computacionais. A tecnologia AutoNAC atravessa o vasto espaço de pesquisa arquitetônica em busca de arquiteturas que equilibrem latência (tempo necessário para receber resultados de inferência) e taxa de transferência (quadros de imagem processados dentro de um período de tempo específico).

A robustez do modelo YOLO-NAS é evidente em sua resiliência a mudanças nos dados de entrada, capacidade de lidar com ruído ou incerteza e manter altas taxas de precisão mesmo durante a quantização pós-treinamento. Além disso, os princípios de detecção de estágio único, divisão de imagem em grade, previsão de caixa delimitadora, previsões em múltiplas escalas e supressão não máxima - que são essenciais para as arquiteturas YOLO - equipam o YOLO-NAS com a robustez para detectar objetos de vários tamanhos em diferentes cenários de forma eficaz.

E agora atingimos a ‘Fronteira da Eficiência’. A fronteira de eficiência refere-se ao espaço de busca que cobre a arquitetura que apresenta um equilíbrio ideal de latência, taxa de transferência e precisão. AutoNAC facilita a descoberta de novas arquiteturas de redes neurais considerando a disponibilidade de hardware, metas de desempenho, quantização, etc. A fronteira de eficiência contém as variantes YOLO-NAS YOLO-NAS-S, YOLO-NAS-M e YOLO-NAS-L; essas variantes de modelo atendem a diferentes restrições computacionais e de hardware, introduzindo escalabilidade baseada em recursos computacionais ao modelo YOLO-NAS.

## Mecanismo de Atenção

Como praticante de aprendizado de máquina, você provavelmente está familiarizado com a técnica de aprendizado de máquina: Mecanismo de atenção, popularizada pelo artigo que apresentou a rede neural Transformer: Atenção é tudo que você precisa.

A arquitetura YOLO-NAS incorpora o mecanismo de atenção e o aproveita para focar seletivamente em certas partes de uma imagem contendo objeto(s) alvo relevante(s) para o domínio do problema ou caso de uso.

A incorporação do mecanismo de atenção do YOLO-NAS em sua arquitetura permite a priorização de áreas de uma imagem que contém um objeto alvo, reduzindo efetivamente a influência de informações irrelevantes, como objetos não alvo e fundo da imagem. Esta aplicação de atenção refina o foco do modelo e aumenta significativamente suas capacidades de detecção de objetos.

## Como o YOLO-NAS foi treinado?

YOLO-NAS passa por um processo de treinamento multifásico que envolve pré-treinamento no conjunto de dados Objects365, utilizando o conjunto de dados COCO para gerar dados pseudo-rotulados e incorporando técnicas de Destilação de Conhecimento (KD) e Perda Focal de Distribuição (DFL).

O pré-treinamento no Objects365, que consiste em 2 milhões de imagens e 365 categorias, dura de 25 a 40 épocas em 8 GPUs NVIDIA RTX A5000. O conjunto de dados COCO fornece 123 mil imagens não rotuladas adicionais, que são usadas para gerar dados pseudo-rotulados para treinar o modelo.

A técnica KD é aplicada adicionando um termo à função de perda, permitindo que a rede de alunos imite tanto a classificação quanto as previsões DFL da rede de professores. Enquanto isso, o DFL é usado para discretizar as previsões de caixa em valores finitos, aprendendo a regressão de caixa como uma tarefa de classificação e prevendo distribuições sobre esses valores, que são então convertidos em previsões finais por meio de uma soma ponderada.

Esses métodos de treinamento permitem que o YOLO-NAS alcance alta precisão e capacidades superiores de detecção de objetos.

```python
INPUT_PATH = path_imgs
OUTPUT_PATH = '/content/letuce_data'
def process_data(data, data_type='train'):
    for image_name in data['Image_ID'].unique():
        new_data = data[data['Image_ID'] == image_name]
        for _, row in new_data.iterrows():
            classes = 0
            yolo_data = []
            x = row['x']
            y = row['y']
            w = row['w']
            h = row['h']
            x_center = x + w / 2
            y_center = y + h / 2

            x_center /= 810
            y_center /= 1080
            w /= 810
            h /= 1080
            yolo_data.append([classes, x_center, y_center, w, h])

        yoy_data = np.array(yolo_data)
        np.savetxt(
            os.path.join(OUTPUT_PATH, f"{data_type}/labels/{image_name.split('.')[0]}.txt"),
            yolo_data,
            fmt = ["%d", "%f", "%f", "%f", "%f"]
        )
        shutil.copyfile(
            os.path.join(INPUT_PATH, f"{image_name}"),
            os.path.join(OUTPUT_PATH, f"{data_type}/images/{image_name}")
        )

process_data(df_train, data_type='train')
process_data(df_test, data_type='test')
```

```python
class config:
    #trainer params
    CHECKPOINT_DIR = 'checkpoints' #specify the path you want to save checkpoints to
    EXPERIMENT_NAME = 'cars-from-above' #specify the experiment name

    #dataset params
    DATA_DIR = '/content/letuce_data' #parent directory to where data lives

    TRAIN_IMAGES_DIR = 'train/images' #child dir of DATA_DIR where train images are
    TRAIN_LABELS_DIR = 'train/labels' #child dir of DATA_DIR where train labels are

    VAL_IMAGES_DIR = 'test/images' #child dir of DATA_DIR where validation images are
    VAL_LABELS_DIR = 'test/labels' #child dir of DATA_DIR where validation labels are

    CLASSES = ['Letuce'] #what class names do you have

    NUM_CLASSES = len(CLASSES)

    #dataloader params - you can add whatever PyTorch dataloader params you have
    #could be different across train, val, and test
    DATALOADER_PARAMS={
    'batch_size':16,
    'num_workers':2
    }

    # model params
    MODEL_NAME = 'yolo_nas_l' # choose from yolo_nas_s, yolo_nas_m, yolo_nas_l
    PRETRAINED_WEIGHTS = 'coco' #only one option here: coco
```

```python
trainer = Trainer(experiment_name=config.EXPERIMENT_NAME, ckpt_root_dir=config.CHECKPOINT_DIR)
```

```python
train_data = coco_detection_yolo_format_train(
    dataset_params={
        'data_dir': config.DATA_DIR,
        'images_dir': config.TRAIN_IMAGES_DIR,
        'labels_dir': config.TRAIN_LABELS_DIR,
        'classes': config.CLASSES
    },
    dataloader_params=config.DATALOADER_PARAMS
)

val_data = coco_detection_yolo_format_val(
    dataset_params={
        'data_dir': config.DATA_DIR,
        'images_dir': config.VAL_IMAGES_DIR,
        'labels_dir': config.VAL_LABELS_DIR,
        'classes': config.CLASSES
    },
    dataloader_params=config.DATALOADER_PARAMS
)
```

```python
train_data.dataset.plot()
```

```python
model = models.get(config.MODEL_NAME,
                   num_classes=config.NUM_CLASSES,
                   pretrained_weights=config.PRETRAINED_WEIGHTS
                   )
```

```python
train_params = {
    # ENABLING SILENT MODE
    "average_best_models":True,
    "warmup_mode": "linear_epoch_step",
    "warmup_initial_lr": 1e-6,
    "lr_warmup_epochs": 3,
    "initial_lr": 5e-4,
    "lr_mode": "cosine",
    "cosine_final_lr_ratio": 0.1,
    "optimizer": "Adam",
    "optimizer_params": {"weight_decay": 0.0001},
    "zero_weight_decay_on_bias_and_bn": True,
    "ema": True,
    "ema_params": {"decay": 0.9, "decay_type": "threshold"},
    # ONLY TRAINING FOR 10 EPOCHS FOR THIS EXAMPLE NOTEBOOK
    "max_epochs": 30,
    "mixed_precision": True,
    "loss": PPYoloELoss(
        use_static_assigner=False,
        # NOTE: num_classes needs to be defined here
        num_classes=config.NUM_CLASSES,
        reg_max=16
    ),
    "valid_metrics_list": [
        DetectionMetrics_050(
            score_thres=0.1,
            top_k_predictions=300,
            # NOTE: num_classes needs to be defined here
            num_cls=config.NUM_CLASSES,
            normalize_targets=True,
            post_prediction_callback=PPYoloEPostPredictionCallback(
                score_threshold=0.01,
                nms_top_k=1000,
                max_predictions=300,
                nms_threshold=0.7
            )
        )
    ],
    "metric_to_watch": 'mAP@0.50'
}
```

```python
trainer.train(model=model,
              training_params=train_params,
              train_loader=train_data,
              valid_loader=val_data)
```

Now that training is complete, you need to get the best trained model.

You used checkpoint averaging so the following code will use weights averaged across training runs.

If you want to use the best weights, or weights from the last epoch you'd use one of the following in the code below:

best weights: checkpoint_path = os.path.join(config.CHECKPOINT_DIR, config.EXPERIMENT_NAME, ckpt_best.pth)

```python
best_model = models.get(config.MODEL_NAME,
                        num_classes=config.NUM_CLASSES,
                        checkpoint_path=os.path.join(config.CHECKPOINT_DIR, config.EXPERIMENT_NAME, 'ckpt_best.pth'))
```

```python
trainer.test(model=best_model,
            test_loader=val_data,
            test_metrics_list=DetectionMetrics_050(score_thres=0.1,
                                                   top_k_predictions=300,
                                                   num_cls=config.NUM_CLASSES,
                                                   normalize_targets=True,
                                                   post_prediction_callback=PPYoloEPostPredictionCallback(score_threshold=0.01,
                                                                                                          nms_top_k=1000,
                                                                                                          max_predictions=300,
                                                                                                          nms_threshold=0.7)))
```
